### Understanding Student Drinking Habits, Lifestyle Patterns, and Their Relationship with Academic Performance

**Dataset:** Student Alcohol Consumption Dataset (`student-mat.csv`) (`student-por.csv`)

### 1. Business Objective

Alcohol consumption among students is an important social and educational concern. Students' lifestyles, social interactions, family environments, and personal habits may be associated with their overall well-being and academic performance.

This dataset provides information about students enrolled in mathematics courses at two secondary schools in Portugal. It includes details such as age, gender, study time, family relationships, social activities, weekday and weekend alcohol consumption, and academic grades.

**Aim:**

> To analyze student alcohol consumption patterns and investigate how demographic, social, and family-related factors are associated with drinking habits and academic performance. The objective is to identify which student groups report higher alcohol consumption, understand the differences between weekday and weekend drinking habits, and explore the relationship between alcohol consumption and academic performance.

In short, this project treats the dataset like a **student lifestyle and academic performance research report** — an analysis that schools, educational institutions, counselors, and policymakers could use to better understand student behavior, identify potential risk factors, and develop awareness programs to promote healthier lifestyles and academic well-being.

### Analysis 1: Age Distribution

```python
ax = sns.countplot(
    data=math,
    x='age',
    color='#7B2D43',
    edgecolor='#F7F3EF'
)

plt.title('Age Distribution of Mathematics Students',
          fontsize=15, fontweight='bold')

plt.xlabel('Student Age', fontsize=11)
plt.ylabel('Number of Students', fontsize=11)

# Add values above bars
for container in ax.containers:
    ax.bar_label(container, padding=3, fontsize=9)

sns.despine()
plt.tight_layout()
plt.show()
```

`Insight:` The Mathematics dataset primarily represents students between 15 and 18 years of age.

```python
ax = sns.countplot(
    data=portuguese,
    x='age',
    color='#52677D',
    edgecolor='#F7F3EF'
)

plt.title('Age Distribution of Portuguese Students',
          fontsize=15, fontweight='bold')

plt.xlabel('Student Age', fontsize=11)
plt.ylabel('Number of Students', fontsize=11)

for container in ax.containers:
    ax.bar_label(container, padding=3, fontsize=9)

sns.despine()
plt.tight_layout()
plt.show()
```

`Insight:` The age distribution of Portuguese students is similar to Mathematics, with most students concentrated in the 16–17 age range.

### Analysis 2: Gender Distribution

```python
gender_count = math['sex'].value_counts()

plt.pie(
    gender_count,
    labels=gender_count.index,
    autopct='%1.1f%%',
    pctdistance=0.8,
    startangle=90,
    colors=['#7B2D43', '#D4A373'],
    wedgeprops={'width': 0.4, 'edgecolor': '#F7F3EF'},
    textprops={'fontsize': 11}
)

plt.title('Gender Distribution - Mathematics Students',
          fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```

`Insight:` The Mathematics dataset contains a slightly higher proportion of female students, but both genders are well represented.

```python
gender_count = portuguese['sex'].value_counts()

plt.pie(
    gender_count,
    labels=gender_count.index,
    autopct='%1.1f%%',
    pctdistance=0.8,
    startangle=90,
    colors=['#52677D', '#C06C84'],
    wedgeprops={'width': 0.4, 'edgecolor': '#F7F3EF'},
    textprops={'fontsize': 11}
)

plt.title('Gender Distribution - Portuguese Students',
          fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```

`Insight:` Female students form the majority of the Portuguese dataset, indicating a higher representation compared with male students.

### Analysis 3: Weekday vs Weekend Alcohol Consumption

```python
alcohol_math = pd.DataFrame({
    'Weekday': math['Dalc'].value_counts().sort_index(),
    'Weekend': math['Walc'].value_counts().sort_index()
})

alcohol_math.plot(
    kind='bar',
    figsize=(10, 5),
    color=['#7B2D43', '#D4A373'],
    edgecolor='#F7F3EF',
    width=0.75
)

plt.title('Weekday vs Weekend Alcohol Consumption - Mathematics',
          fontsize=14, fontweight='bold')

plt.xlabel('Alcohol Consumption Level (1 = Very Low, 5 = Very High)')
plt.ylabel('Number of Students')
plt.xticks(rotation=0)
plt.legend(title='Consumption Period')

plt.tight_layout()
plt.show()
```

`Insight:` Weekend alcohol consumption is more common at higher consumption levels than weekday consumption among Mathematics students.

```python
alcohol_por = pd.DataFrame({
    'Weekday': portuguese['Dalc'].value_counts().sort_index(),
    'Weekend': portuguese['Walc'].value_counts().sort_index()
})

alcohol_por.plot(
    kind='bar',
    figsize=(10, 5),
    color=['#52677D', '#C06C84'],
    edgecolor='#F7F3EF',
    width=0.75
)

plt.title('Weekday vs Weekend Alcohol Consumption - Portuguese',
          fontsize=14, fontweight='bold')

plt.xlabel('Alcohol Consumption Level (1 = Very Low, 5 = Very High)')
plt.ylabel('Number of Students')
plt.xticks(rotation=0)
plt.legend(title='Consumption Period')

plt.tight_layout()
plt.show()
```

`Insight:` Both datasets indicate a clear difference between weekday and weekend drinking patterns, with weekend consumption generally being higher.

`Insight:` Portuguese students show a concentration of average academic scores, with relatively fewer students achieving extremely high or low grades.

### Analysis 5: Relationship Between Alcohol Consumption and Academic Performance

```python
weekday_math = math.groupby('Dalc')['G3'].mean().reset_index()

ax = sns.barplot(
    data=weekday_math,
    x='Dalc',
    y='G3',
    color='#7B2D43',
    edgecolor='#F7F3EF'
)

plt.title('Weekday Alcohol Consumption vs Final Grades - Mathematics',
          fontsize=14, fontweight='bold')

plt.xlabel('Weekday Alcohol Consumption Level')
plt.ylabel('Average Final Grade (G3)')

for container in ax.containers:
    ax.bar_label(container, fmt='%.2f', padding=3)

plt.tight_layout()
plt.show()
```

`Insight:` The Mathematics dataset does not show a clear linear relationship between weekday alcohol consumption and final academic grades.

```python
weekend_math = math.groupby('Walc')['G3'].mean().reset_index()

ax = sns.barplot(
    data=weekend_math,
    x='Walc',
    y='G3',
    color='#D4A373',
    edgecolor='#F7F3EF'
)

plt.title('Weekend Alcohol Consumption vs Final Grades - Mathematics',
          fontsize=14, fontweight='bold')

plt.xlabel('Weekend Alcohol Consumption Level')
plt.ylabel('Average Final Grade (G3)')

for container in ax.containers:
    ax.bar_label(container, fmt='%.2f', padding=3)

plt.tight_layout()
plt.show()
```

`Insight:` Weekend alcohol consumption shows some variation in academic performance, but the relationship is not strongly consistent.

```python
weekday_por = portuguese.groupby('Dalc')['G3'].mean().reset_index()

ax = sns.barplot(
    data=weekday_por,
    x='Dalc',
    y='G3',
    color='#52677D',
    edgecolor='#F7F3EF'
)

plt.title('Weekday Alcohol Consumption vs Final Grades - Portuguese',
          fontsize=14, fontweight='bold')

plt.xlabel('Weekday Alcohol Consumption Level')
plt.ylabel('Average Final Grade (G3)')

for container in ax.containers:
    ax.bar_label(container, fmt='%.2f', padding=3)

plt.tight_layout()
plt.show()
```

`Insight:` The Portuguese dataset shows a noticeable tendency toward lower academic performance at higher weekday alcohol consumption levels, although the pattern is not perfectly linear.

```python
weekend_por = portuguese.groupby('Walc')['G3'].mean().reset_index()

ax = sns.barplot(
    data=weekend_por,
    x='Walc',
    y='G3',
    color='#C06C84',
    edgecolor='#F7F3EF'
)

plt.title('Weekend Alcohol Consumption vs Final Grades - Portuguese',
          fontsize=14, fontweight='bold')

plt.xlabel('Weekend Alcohol Consumption Level')
plt.ylabel('Average Final Grade (G3)')

for container in ax.containers:
    ax.bar_label(container, fmt='%.2f', padding=3)

plt.tight_layout()
plt.show()
```

`Insight:` Higher weekend alcohol consumption is associated with lower average final grades among Portuguese students.

### Analysis 9: Final Grade by Subject

```python
comparison = pd.DataFrame({
    'Subject': ['Mathematics', 'Portuguese'],
    'Average Grade': [math['G3'].mean(),
                      portuguese['G3'].mean()]
})

sns.barplot(data=comparison, x='Subject', y='Average Grade',
            hue='Subject', palette=['#7B2D43', '#D4A373'],
            legend=False)

plt.title('Average Final Grade by Subject')
plt.ylabel('Average G3 Grade')
plt.show()
```

`Insight:` Portuguese students demonstrate higher average final grades than Mathematics students in the analyzed datasets.

### Analysis 10: Study Time vs Alcohol Consumption – Heatmap

```python
table = pd.crosstab(math['studytime'],
                    math['Walc'], normalize='index')

sns.heatmap(table, annot=True, cmap='RdPu', fmt='.2f')

plt.title('Study Time vs Weekend Alcohol Consumption')
plt.xlabel('Weekend Alcohol Level')
plt.ylabel('Study Time Level')
plt.show()
```

`Insight:` The heatmap suggests that students who spend more time studying generally report lower weekend alcohol consumption. However, this represents an association rather than a causal relationship.

### Analysis 11: Age vs Alcohol Consumption on Weekday

```python
sns.violinplot(data=math, x='Dalc', y='age',color='#D4A373')

plt.title('Age Distribution by Weekday Alcohol Level')
plt.xlabel('Weekday Alcohol Level')
plt.ylabel('Age')
plt.show()
```

`Insight:` Older students appear across higher weekday alcohol consumption categories, but the chart does not establish a strong relationship between age and alcohol consumption.
